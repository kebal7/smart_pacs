```python
from pydicom import dcmread
from pydicom.data import get_testdata_file
import matplotlib.pyplot as plt
import numpy as np

# Load DICOM file
filename = get_testdata_file("CT_small.dcm")
ds = dcmread(filename)

# Access pixel data as a NumPy array
pixel_array = ds.pixel_array  # This automatically converts PixelData to an array

print("Shape of image:", pixel_array.shape)
print("Data type:", pixel_array.dtype)

# Show the image
plt.imshow(pixel_array, cmap='bone')
plt.axis('off')
plt.show()
```
